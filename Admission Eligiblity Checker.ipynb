```python
print("===== University Admission Eligibility Checker =====")

student_name = input("Enter student name: ")
age = int(input("Enter your age: "))
matric_marks = int(input("Enter matric marks out of 1100: "))
inter_marks = int(input("Enter intermediate marks out of 1100: "))
entry_test_marks = int(input("Enter entry test marks out of 100: "))
program = input("Enter program name BSCS/BBA: ")
```

```python
print("\n===== Admission Report =====")
print("Student Name:", student_name)
print("Age:", age)
print("Matric Marks:", matric_marks)
print("Intermediate Marks:", inter_marks)
print("Entry Test Marks:", entry_test_marks)
print("Selected Program:", program)
```

```python

if (
    matric_marks >= 0
    and matric_marks <= 1100
    and inter_marks >= 0
    and inter_marks <= 1100
    and entry_test_marks >= 0
    and entry_test_marks <= 100
):
    print("Marks Status: Valid")
    matric_percentage = (matric_marks / 1100) * 100
    inter_percentage = (inter_marks / 1100) * 100
    print("Matric Percentage:", matric_percentage)
    print("Intermediate Percentage:", inter_percentage)

    # Checking age eligibility
    if age >= 16 and age <= 30:
        print("Age Status: Eligible")

        # BSCS admission condition
        if program == "BSCS":
            if inter_percentage >= 60 and entry_test_marks >= 50:
                print("Admission Status: Approved")
                print("Department: Computer Science")
                if inter_percentage >= 80 and entry_test_marks >= 80:
                    print("Merit Category: High Merit")
                else:
                    print("Merit Category: Normal Merit")
            else:
                print("Admission Status: Rejected")
                if inter_percentage < 60 or entry_test_marks < 50:
                    print(
                        "Reason: Low intermediate percentage or low entry test marks"
                    )

       
        elif program == "BBA":
            if inter_percentage >= 50 and entry_test_marks >= 40:
                print("Admission Status: Approved")
                print("Department: Business Administration")
                if inter_percentage >= 75 or entry_test_marks >= 75:
                    print("Merit Category: Good Merit")
                else:
                    print("Merit Category: Normal Merit")
            else:
                print("Admission Status: Rejected")
                if inter_percentage < 50 or entry_test_marks < 40:
                    print(
                        "Reason: Low intermediate percentage or low entry test marks"
                    )
        else:
            print("Admission Status: Rejected")
            print("Reason: Program name is incorrect")
    else:
        print("Age Status: Not Eligible")
        print("Admission Status: Rejected")
else:
    print("Marks Status: Invalid")
    print("Admission Status: Rejected")
```

```python
# Extra relational check
if entry_test_marks == 100:
    print("Entry Test Result: Perfect Score")
else:
    if entry_test_marks != 100:
        print("Entry Test Result: Not a perfect score")

print("\n===== Program End =====")
```
